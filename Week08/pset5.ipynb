## Q4

```python
import numpy as np
import pandas as pd
from scipy.stats import norm

df = pd.read_csv('./data/ames_prices.csv')
x = (df['Central.Air'] == 'Y').astype(int)
n = len(x)
p_hat = x.mean()
print(f'n = {n}, p_hat = {p_hat:.4f}')
```

```python
np.random.seed(1)
R = 2_000
results = []
for t in range(R):
    idx = np.random.randint(0, n, n)
    results.append(x.iloc[idx].mean())
results = np.array(results)
print(f'seed = 1, R = {R}')
```

```python
p_0 = .90
se = np.std(results, ddof=1)
z = (p_hat - p_0) / se
p = 2 * norm.cdf(-np.abs(z))
row = {'variable': 'Central.Air', 'estimate': p_hat, 'SE': se, 'z': z, 'p': p}
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    row[f'normal {conf}%'] = f'({p_hat - c*se:.4f}, {p_hat + c*se:.4f})'
    row[f'contains p_0 {conf}%'] = p_hat - c*se <= p_0 <= p_hat + c*se
row['significant at'] = ', '.join(f'{conf}%' for conf, alpha in [(90, .10), (95, .05), (99, .01)] if p < alpha) or 'none'

ci_table = pd.DataFrame([row]).set_index('variable')
ci_table.round(3)
```

About 93.3% of homes in the sample have central air. The 95% CI is (92.4, 94.2), so the true proportion is probably in that range. 0.90 is outside all three intervals and p is approx 0, so we reject $H_0$ at 90, 95, and 99%. The share of ames homes with central air is higher than 90%.

```python
np.random.seed(1)
R = 2_000
results = []
for t in range(R):
    idx = np.random.randint(0, n, n)
    results.append(np.corrcoef(x.iloc[idx], y.iloc[idx])[0, 1])
results = np.array(results)
print(f'seed = 1, R = {R}')
```

```python
rho_0 = 0
se = np.std(results, ddof=1)
z = (rho_hat - rho_0) / se
p = 2 * norm.cdf(-np.abs(z))
row = {'variable': 'corr(area, price)', 'estimate': rho_hat, 'SE': se, 'z': z, 'p': p}
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    row[f'normal {conf}%'] = f'({rho_hat - c*se:.4f}, {rho_hat + c*se:.4f})'
    row[f'contains rho_0 {conf}%'] = rho_hat - c*se <= rho_0 <= rho_hat + c*se
row['significant at'] = ', '.join(f'{conf}%' for conf, alpha in [(90, .10), (95, .05), (99, .01)] if p < alpha) or 'none'

ci_table = pd.DataFrame([row]).set_index('variable')
ci_table.round(3)
```

The correlation between arcinh of area and price is 0.723, so bigger houses tend to sell for more. 0 is not included in all three intervals and p is close to 0, so we reject $H_0$ at 90, 95, and 99%. 

If area and price were resample separately, each house's area would pair with a other random house's price, so we don't see a real relationship, and every bootstrap correlation would be close to 0. The standard error and CIs would not describe a real relationship.

```python
df = pd.read_csv('./data/metabric.csv')
x = df['Overall Survival (Months)']
n = len(x)
mu_hat = x.mean()
print(f'n = {n}, mu_hat = {mu_hat:.4f}')
```

```python
np.random.seed(1)
R = 2_000
results = []
for t in range(R):
    idx = np.random.randint(0, n, n)
    results.append(x.iloc[idx].mean())
results = np.array(results)
print(f'seed = 1, R = {R}')
```

```python
mu_0 = 120
se = np.std(results, ddof=1)
z = (mu_hat - mu_0) / se
p = 2 * norm.cdf(-np.abs(z))
row = {'variable': 'overall Survival (Months)', 'estimate': mu_hat, 'SE': se, 'z': z, 'p': p}
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    row[f'normal {conf}%'] = f'({mu_hat - c*se:.2f}, {mu_hat + c*se:.2f})'
    row[f'normal {conf}% (years)'] = f'({(mu_hat - c*se)/12:.2f}, {(mu_hat + c*se)/12:.2f})'
    row[f'contains mu_0 {conf}%'] = mu_hat - c*se <= mu_0 <= mu_hat + c*se
row['significant at'] = ', '.join(f'{conf}%' for conf, alpha in [(90, .10), (95, .05), (99, .01)] if p < alpha) or 'none'

ci_table = pd.DataFrame([row]).set_index('variable')
ci_table.round(3)
```

The mean recorded time is 129.1 months. 120 months is excluded in all three intervals and $p$ is close to $0$, so we reject $H_0$ at 90, 95, and 99%. The mean recorded time is longer than 10 years.

This is not the average lifetime after diagnosis. 576 of the 1343 were still living, so each time is how long they were followed, not until they died. Their survival times are longer than recorded, so the true mean lifetime is be higher than our estimate.
