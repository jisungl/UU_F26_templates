```python
# new
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("./data/metabric.csv")

fig, ax = plt.subplots(figsize=(7, 4.5))
sns.ecdfplot(data=df, x="Overall Survival (Months)", color="black", ax=ax)
ax.set_ylabel("ECDF")
ax.set_title("ECDF of Overall Survival (Months)")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
for label, style in [("NO", "--"), ("YES", "-")]:
    sub = df[df["Chemotherapy"] == label]
    sns.ecdfplot(data=sub, x="Overall Survival (Months)",
                 color="black", linestyle=style, label=label, ax=ax)

ax.legend(title="Chemotherapy")
ax.set_ylabel("ECDF")
ax.set_title("ECDF of Overall Survival (Months) by Chemotherapy")
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
